```python
import pandas as pd

# 1. Create decade column
names['decade'] = (names['year'] // 10) * 10

# 2. Aggregate total births by name, decade, and sex
grouped = names.groupby(['name', 'decade', 'sex'])['births'].sum().unstack(fill_value=0)
grouped['total'] = grouped['F'] + grouped['M']

# Filter names with total births >= 500 in both 1900s and 2000s
total_by_decade = grouped['total'].unstack(fill_value=0)
valid_names = total_by_decade[(total_by_decade[1900] >= 500) & (total_by_decade[2000] >= 500)].index

# 3. Calculate female ratio (F / Total) for each decade
female_ratio = (grouped['F'] / grouped['total']).unstack(fill_value=0)
female_ratio_filtered = female_ratio.loc[valid_names]

# 4. Find gender-swapped names (female ratio < 20% in 1900s and > 80% in 2000s)
gender_swapped_names = female_ratio_filtered[
    (female_ratio_filtered[1900] < 0.20) &
    (female_ratio_filtered[2000] > 0.80)
]

# Display results
print("Gender-swapped names (Male in 1900s to Female in 2000s):")
print(gender_swapped_names[[1900, 2000]])
```

```python
import pandas as pd
import numpy as np
import requests
import aiohttp
import asyncio
import time
import matplotlib.pyplot as plt
import seaborn as sns
from io import StringIO

# GitHub API
url = "https://raw.githubusercontent.com/wesm/pydata-book/3rd-edition/datasets/babynames/yob1880.txt"

response = requests.get(api_url)
files = response.json()

# Select only txt files from 1880 to 2010
txt_files = [
    file['name'] for file in files
    if file['name'].endswith('.txt')
    and file['name'][3:7].isdigit()
    and 1880 <= int(file['name'][3:7]) <= 2010
]

txt_files = sorted(txt_files)

print("Number of files:", len(txt_files))
print("First files:", txt_files[:5])
print("Last files:", txt_files[-5:])
```

```python
base_url = "https://raw.githubusercontent.com/wesm/pydata-book/3rd-edition/datasets/babynames/"

start_time = time.time()

data_list = []

for filename in txt_files:
    year = int(filename[3:7])

    df = pd.read_csv(
        base_url + filename,
        header=None,
        names=['name', 'sex', 'births']
    )

    df['year'] = year
    data_list.append(df)

data_sync = pd.concat(data_list, ignore_index=True)

sync_time = time.time() - start_time

print("Sequential loading time:", round(sync_time, 2), "seconds")
print(data_sync.head())
```

```python
async def download_file(session, filename):
    year = int(filename[3:7])
    url = base_url + filename

    async with session.get(url) as response:
        text = await response.text()

        df = pd.read_csv(
            StringIO(text),
            header=None,
            names=['name', 'sex', 'births']
        )

        df['year'] = year
        return df


async def download_all():
    connector = aiohttp.TCPConnector(limit=15)

    async with aiohttp.ClientSession(connector=connector) as session:
        tasks = [
            download_file(session, filename)
            for filename in txt_files
        ]

        results = await asyncio.gather(*tasks)

    return pd.concat(results, ignore_index=True)


start_time = time.time()

data_async = await download_all()

async_time = time.time() - start_time

print("Parallel loading time:", round(async_time, 2), "seconds")

print("Sequential loading:", round(sync_time, 2), "seconds")
print("Parallel loading:", round(async_time, 2), "seconds")

print(data_async.head())
```

```python
diversity = (
    data.sort_values('prop', ascending=False)
    .groupby(['year', 'sex'])
    .apply(
        lambda group: (group['prop'].cumsum() < 0.5).sum() + 1,
        include_groups=False
    )
    .reset_index(name='names_needed')
)

print(diversity.head())

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=diversity,
    x='year',
    y='names_needed',
    hue='sex'
)

plt.title('Number of Names Needed to Reach 50% of Births')
plt.xlabel('Year')
plt.ylabel('Number of Names')
plt.grid()
plt.show()
```

```python
data['last_letter'] = data['name'].str[-1].str.lower()

last_letters = (
    data.groupby(['year', 'sex', 'last_letter'])['births']
    .sum()
    .reset_index()
)

last_letters['prop'] = last_letters['births'] / last_letters.groupby(
    ['year', 'sex']
)['births'].transform('sum')

# Create separate heatmaps for boys and girls
for sex in ['M', 'F']:

    letter_data = last_letters[last_letters['sex'] == sex]

    letter_pivot = letter_data.pivot(
        index='year',
        columns='last_letter',
        values='prop'
    ).fillna(0)

    plt.figure(figsize=(16, 8))

    sns.heatmap(letter_pivot, cmap='YlOrRd')

    plt.title(f'Popularity of Last Letters - {sex}')
    plt.xlabel('Last Letter')
    plt.ylabel('Year')
    plt.show()
```

### 동기방식과 비동기 방식의 파일 다운로드 코드 실행시간 비교

```python
# @title 동기방식(Syn), Blocking
import time

# 1. 측정 시작
start_time = time.perf_counter()

base_url = "https://raw.githubusercontent.com/wesm/pydata-book/3rd-edition/datasets/babynames/yob{}.txt"

# 1880년부터 2011년까지 전체 데이터 읽어오기
dfs = []
names=["name", "sex", "births"]

for year in range(1880, 2011):
    url = base_url.format(year)
    # print(url)
    # pandas는 웹 URL을 직접 읽기
    df = pd.read_csv(url, names= names)
    df["year"] = year
    dfs.append(df)

names = pd.concat(dfs, ignore_index=True)

# 2. 측정 종료
end_time = time.perf_counter()

# 3. 소요 시간 계산
execution_time = end_time - start_time
print(f"전체 코드 실행 시간: {execution_time:.2f}초")
```

```python
# @title 비동기방식(Asyn), Non-Blocking
import asyncio
import io
import aiohttp

# 1. 측정 시작
start_time = time.perf_counter()

base_url = (
    "https://raw.githubusercontent.com/wesm/pydata-book/3rd-edition/datasets/babynames/yob{}.txt"
)
names = ["name", "sex", "births"]


async def fetch(session, year):
    url = base_url.format(year)
    async with session.get(url) as response:
        # 텍스트 데이터를 메모리 바이너리로 변환 후 read_csv 처리
        content = await response.read()
        df = pd.read_csv(io.BytesIO(content), names=names)
        df["year"] = year
        return df


async def main():
    async with aiohttp.ClientSession() as session:
        # 1880년부터 2010년(2011 전)까지 동시 요청 생성
        tasks = [fetch(session, year) for year in range(1880, 2011)]
        dfs = await asyncio.gather(*tasks)
        return pd.concat(dfs, ignore_index=True)


# 실행 (Jupyter Notebook / Google Colab 환경)
names = await main()
# 일반 Python 스크립트(.py 파일) 실행 시 아래 주석 해제:
# users = asyncio.run(main())

# 2. 측정 종료
end_time = time.perf_counter()

# 3. 소요 시간 계산
execution_time = end_time - start_time
print(f"전체 코드 실행 시간: {execution_time:.2f}초")
```

```python
def get_quantile_count(group, q=0.5):
    group = group.sort_values("prop", ascending=False)
    return group.prop.cumsum().searchsorted(q) + 1

diversity = top1000.groupby(["year", "sex"]).apply(get_quantile_count)
diversity
```
